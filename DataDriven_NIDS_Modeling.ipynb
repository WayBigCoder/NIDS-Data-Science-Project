  <div class="section-heading">1. Feature Selection</div>
  <p class="paragraph">
    The core of an effective Network Intrusion Detection System lies in its ability to accurately identify patterns indicative of intrusions. This capability hinges on the quality and relevance of the features used in the machine learning models. 
  </p>

  <ul class="paragraph">
    <li><strong>Basic Statistical Features:</strong> mean, max, min, and standard deviation were chosen for their ability to capture the central tendency and dispersion in the data. These statistics are important in understanding the normal behavior of network traffic, thereby making anomalies more detectable.</li>
    <li><strong>Packet Arrival Time Analysis:</strong> For flows with multiple packets, the average, maximum, minimum, and standard deviation of arrival times were calculated. These time-based features help in identifying irregular traffic patterns, which are often a sign of intrusion attempts.</li>
    <li><strong>Flow Duration</strong> </li>
  </ul>

  <p class="paragraph">
    Each of these features contributes uniquely to the model's ability to distinguish between normal and malicious network traffic. For instance, unusual variations in basic statistical features can flag potential threats, while anomalies in flow duration and packet arrival times can identify more sophisticated intrusion tactics.
  </p>
</div>


```python
from typing import List, Literal
import numpy as np

def extract(
        protocol: Literal['UDP', 'TCP'],
        src: str,
        sport: int,
        dst: str,
        dport: int,
        timestamps: np.ndarray,
        sizes: np.ndarray,
    ) -> List[float]:

    # basic statistics
    num_packets = len(timestamps)
    avg_packet_size = np.mean(sizes)
    max_packet_size = np.max(sizes)
    min_packet_size = np.min(sizes)
    std_packet_size = np.std(sizes)
    
    # time-based features
    if num_packets > 1:
        inter_arrival_times = np.diff(timestamps)
        avg_inter_arrival_time = np.mean(inter_arrival_times)
        max_inter_arrival_time = np.max(inter_arrival_times)
        min_inter_arrival_time = np.min(inter_arrival_times)
        std_inter_arrival_time = np.std(inter_arrival_times)
    else:
        # see expalnation in Question 1.1
        avg_inter_arrival_time = 0
        max_inter_arrival_time = 0
        min_inter_arrival_time = 0
        std_inter_arrival_time = 0
    
    # flow duration
    flow_duration = timestamps[-1] - timestamps[0] if num_packets > 1 else 0

    features = [
        num_packets,
        avg_packet_size,
        max_packet_size,
        min_packet_size,
        std_packet_size,
        avg_inter_arrival_time,
        max_inter_arrival_time,
        min_inter_arrival_time,
        std_inter_arrival_time,
        flow_duration,
    ]
    
    return features

# Set extract method
assignment.extract = extract
```

  <div class="section-heading">2. Data Preprocessing</div>
  <p class="paragraph">
    Effective data preprocessing is a critical step in ensuring the accuracy and efficiency of machine learning models. In our Network Intrusion Detection System project, special attention is given to feature scaling, a process vital for balancing the influence of different features during model training.
  </p>

  <p class="paragraph">
    Our dataset comprises various features, each with distinct value ranges. For example, packet lengths are constrained by the Maximum Transmission Unit (MTU), typically 1500 bytes, creating a range of [0, 1500]. In contrast, the average time between packets might fall within a smaller range, such as [0, 10] seconds. Such disparities in ranges can lead machine learning algorithms to disproportionately weigh features with larger absolute ranges.
  </p>

  <p class="paragraph">
    To address this, we employ <strong>min-max scaling</strong>. This technique rescales each feature to a [0, 1] range, maintaining the relative distances between values. The formula for min-max scaling is:
  </p>


$$z_i = \frac{x_i - \min{(X)}}{\max{(X)} - \min{(X)}}.$$


  <p class="paragraph">
    where z_i is the scaled value, x_i is the original value, and X is the set of all values for the feature.
  </p>

  <p class="paragraph">
    <em>Important:</em> The method should return the scaled matrix, along with the computed minimum and maximum values for each feature in the matrix. This is essential for ensuring that the same scaling parameters are applied both during training and when processing new data in the future.
  </p>
</div>



```python
def scale(
        matrix: np.ndarray,
        minimum: Optional[np.ndarray] = None,
        maximum: Optional[np.ndarray] = None,
    ) -> Tuple[list, list, list]:
    
    # If minimum value is not given, compute the minimum for each feature
    if minimum is None:
        minimum = np.min(matrix, axis=0)
        
    # If maximum value is not given, compute the maximum for each feature
    if maximum is None:
        maximum = np.max(matrix, axis=0)
        
    # Use the minimum and maximum to compute the scaled matrix
    range = np.where((maximum - minimum) == 0, 1, maximum - minimum)
    scaled = (matrix - minimum) / range
        
    return scaled, minimum, maximum
    
    
# Sets scale method
assignment.scale = scale
```

  <div class="section-heading">3. Model/Parameter Selection</div>
  <p class="paragraph">
    With our feature set scaled and ready, the next crucial step is selecting an appropriate anomaly detection algorithm for our Network Intrusion Detection System (NIDS). Given that we only have access to benign data and unknown data, we cannot train a classifier. Therefore, we need to train a novelty/outlier detection algorithm. Specifically, I will use a One-Class SVM.
  </p>

  <p class="paragraph">
    A One-Class SVM distinguishes itself by using a kernel function, denoted as K, to establish a geometric relationship between a feature vector X and the support vectors Y. In our project, we utilize the RBF (Radial Basis Function or Gaussian) kernel, defined as:
  </p>


$$K(X, Y) = e^{-\frac{||x - y||^2}{2\sigma^2}}$$


  <p class="paragraph">
    In this formula, (||x - y||^2) represents the Euclidean distance between vectors (x) and (y), and (sigma) is a free parameter defining the spread of the kernel. This RBF kernel is essential for capturing the non-linear relationships in our data.
  </p>
  <p class="paragraph">
    <em>Note:</em> For more information on the Euclidean distance, refer to <a href="https://

```python
def K(X: np.ndarray, Y: np.ndarray, sigma: float=1) -> float:
    # The code between this text should be implemented
    euclidean_distance = np.sum((X - Y) ** 2, axis=0)

    kernel_func = np.exp(-euclidean_distance / (2 * sigma ** 2))

    return kernel_func
```

A support vector machine (SVM) uses its kernel function to compute a score between a feature vector $X$ and support vector $Y$. You can think of the score as the likelihood **(note that it is not a probability)** that $X$ and $Y$ originate from the same underlying distribution. Because an SVM contains multiple support vectors $Y_i$, we can assign a score to feature vector $X$ based on the likelihood that it shares the same underlying distribution with any of our support vectors $Y_i \in Y$. We compute this overall score by simply adding the (weighted) scores for $X$ with each support vector $Y_i$, which is called the soft score:

$$\text{soft score} = \sum_{i=0}^N w_iK(X, Y_i) \leq \tau$$

We compare this soft score with a threshold $\tau$ to determine if a point falls within or outside of our model. The assignment automatically computes the soft score with all weights as 1 for all given points $X$, support vectors $Y$ and your implemented kernel $K$.

We show the influence of different kernel values on the decision function for random support vectors and datapoints. All datapoints that appear in the red area are considered to fall within the soft hypersphere. All datapoints outside the area are considered anomalous. We see that the sigma changes the shape of the hypershpere and the threshold changes the volume while preserving the shape.

```python
# Plots random kernels and shows the spheres produced by your kernel
assignment.plot_kernels(K, sigmas=[1, 2, 3], thresholds=[3, 2, 1])
```

Later in this project, we will use the `OneClassSVM` implementation from the `scikit-learn` library. This implementation automatically choses optimal values for the weights $w_i$ and threshold $\tau$. However, we can still set two values to influence this decision: `gamma` and `nu`. The `gamma` value determines the size of the kernel, similar to the value of $\sigma$. Additionally, the `nu` value sets an upper bound on the fraction of training errors and a lower bound of the fraction of support vectors. We recall from the lecture that we want to minimize the threshold $\tau$. The `nu` value allows us to adjust the number of support vectors as well as set the allowed number of errors the OneClassSVM can make during training while keeping the $\tau$ value minimal.

  <div class="section-heading">4. Evaluation</div>

Now that we have taken a closer look at One-Class SVMs, we have to think about how we can train, test and evaluate the model. We do this to get an idea of how well the NIDS will perform in a real-world scenario, which allows us to optimise its parameters before deployment and classification of the unknown data. In order to evaluate our model, we require labelled data, i.e. data from the `benign.csv` file. Given that all data is benign, it might seem difficult to assess whether there are malicious network connections. However, we can think of ways to redefine our problem such that we can only use benign data in order to train and test the NIDS.

All data in the `benign.csv` file is labelled with the corresponding application that produced the data. We know that a malicious program is just another application that we have not seen in our benign data. We will leverage this observation to train and test our model.

To this end, we randomly select a couple of applications that we leave out of the training data and only use for the test dataset. These applications simulate the 'unknown' apps or malware that we might encounter in the `unknown.csv` dataset. All other applications will be present in both the train and test set. However, for these applications, we still need to select a portion of the data that we use in the test set, and a portion of the data that we use in the train set. We can define this portion with a `ratio`.

As an example, consider a dataset of 5 applications: $Apps = [\text{Firefox}, \text{Word}, \text{Git}, \text{Excel}, \text{Outlook}]$. Now we randomly select $[\text{Firefox}, \text{Excel}]$ to be our apps for the testing set, meaning $[\text{Word}, \text{Git}, \text{Outlook}]$ are the apps present in both the train and test datasets. Next we define our `ratio` for training items as `0.75`. This means that we select 75% of the flows from $[\text{Word}, \text{Git}, \text{Outlook}]$ as our train data and add the remaining 25% of flows to the test dataset.

```python
def split(
        data: np.ndarray,
        labels: np.ndarray,
        apps_train: List[str],
        apps_test: List[str],
        ratio: float,
    ) -> Tuple[list, list, list, list]:

    train_set = []
    train_labels = []
    delete_indices = []

    # Count the number of rows in the original dataset where labels match those for training
    rows_for_train_set = np.sum(np.isin(labels, apps_train))

    # Number of rows which will be taken from original dataset according to the given ratio
    x = round(ratio * rows_for_train_set) - 1

    # case where no training samples were found; create an empty array with the correct shape.
    if (x == -1):
        x = 0
        train_set = np.zeros((0, data.shape[1]))


    # print("Lenght of labels: ", len(labels))
    # print("Lenght of data: ", len(data))
    # print("How many times 'train labels' appears in original dataset: ", np.sum(np.isin(labels, apps_train)) )
    # print("Number of chosen rows for the train_set", x)
    # print("----------------------------")
        
    # Loop through all labels and add the corresponding data from original dataset to the training set,
    for i, label in enumerate(labels):

        # If the label is in apps_train and we haven't added x rows yet, add to test_set
        if label in apps_train and len(train_set) <= x:
            train_set.append(data[i])
            train_labels.append(labels[i])

            # Track which indices added to the training set to remove it later from original dataset.
            delete_indices.append(i)

    # Delete the test rows from the original dataset
    remaining_data = np.delete(data, delete_indices, axis=0)
    remaining_labels = np.delete(labels, delete_indices, axis=0)

    # The remaining data becomes the testing set
    test_set = remaining_data 
    test_labels = remaining_labels

    return train_set, train_labels, test_set, test_labels

    
# Sets split method
assignment.split = split
```

Now that we have split our data into training and testing sets with the corresponding labels, we want to compute some metrics over a prediction. We recall from the lecture that most metrics are based on the True Positive (TP), False Positive (FP), True Negative (TN) and False Negative values (FN).

 * True Positives (TP), the amount of samples predicted as *unknown* (-1), that should be labelled as *unknown* (-1).
 * True Negatives (TN), the amount of samples predicted as *known* (+1), that should be labelled as *known* (+1).
 * False Positives (FP), the amount of samples predicted as *unknown* (-1), that should be labelled as *known* (+1).
 * False Negatives (FN), the amount of samples predicted as *known* (+1), that should be labelled as *unknown* (-1).
 
Or in a diagram:

| $ $                | Actual Unknown      | Actual Known        |
|--------------------|---------------------|---------------------|
| Predicted Unknown  | True Positive (TP)  | False Positive (FP) |
| Predicted Known    | False Negative (FN) | True Negative (TN)  |

**Question 4.2.** Given these definitions of TP, TN, FP, FN, implement the methods that compute each value given the predictions (`y_pred`) and actual labels (`y_true`).

We use your implementation provided above to compute several performance metrics. There are various metrics that are used widely in the evaluation of machine learning techniques. We recall 5 of these from the lecture:

The True Positive Rate (TPR), also called sensitivity, measures the proportion of actual positives that are correctly identified as such.
$$\text{TPR} = \frac{\text{TP}}{\text{TP}+\text{FN}} = 1-\text{FNR}$$

The True Negative Rate (TNR), also called specificity, measures the proportion of actual negatives that are correctly identified as such.
$$\text{TNR} = \frac{\text{TN}}{\text{TN}+\text{FP}} = 1-\text{FPR}$$

The False Positive Rate (FPR), measures the proportion of actual negatives that are identified as positives. If a NIDS has a high FPR, it will not find malicious software, making it useless.
$$\text{FPR} = \frac{\text{FP}}{\text{TN}+\text{FP}} = 1-\text{TNR}$$

The False Negative Rate (FNR), measures the proportion of actual positives that are identified as negatives. If a NIDS has a high FNR, it will raise alarms for benign network traffic, causing a lot of frustration for the people on the network.
$$\text{FNR} = \frac{\text{FN}}{\text{TP}+\text{FN}} = 1-\text{TPR}$$

The Accuracy (ACC) combines the TPR and FPR to get a complete overview of the system performance. An accuracy of 1.0 means everything is correctly classified. An accuracy of 0.0 means nothing is correctly classified.
$$\text{ACC} = \frac{\text{TP}+\text{TN}}{\text{TP}+\text{TN}+\text{FP}+\text{FN}}$$

Additionally, you will also often encounter the [Precision, recall](https://en.wikipedia.org/wiki/Precision_and_recall) and [F1-score](https://en.wikipedia.org/wiki/F-score). The precision gives the ratio of 'how many selected items are relevant' and the recall gives the ratio of 'how many relevant items are selected'. The F1-score gives the harmonic mean between precision and recall and is often used in place of the accuracy. For reference, the formulas for all three metrics are given below.

$$\text{Precision} = \frac{\text{TP}}{\text{TP}+\text{FP}}$$

$$\text{Recall} = \text{TPR} = \frac{\text{TP}}{\text{TP}+\text{FN}}$$

$$\text{F1-score} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}} = \frac{2 \text{TP}}{2 \text{TP} + \text{FP} + \text{FN}}$$

We compute these metrics based on your True Positive, True Negative, False Positive, and False Negative implementations.

```python
def TP(y_true: np.ndarray, y_pred: np.ndarray) -> int:
    return np.sum((y_pred == -1) & (y_true == -1))

def TN(y_true: np.ndarray, y_pred: np.ndarray) -> int:
    return np.sum((y_pred == 1) & (y_true == 1))

def FP(y_true: np.ndarray, y_pred: np.ndarray) -> int:
    return np.sum((y_pred == -1) & (y_true == 1))

def FN(y_true: np.ndarray, y_pred: np.ndarray) -> int:
    return np.sum((y_pred == 1) & (y_true == -1))
    
    
# Sets True/False Positive/Negative methods
assignment.TP = TP
assignment.TN = TN
assignment.FP = FP
assignment.FN = FN
```
